```python
class Solution:
    def wordPattern(self, pattern: str, s: str) -> bool:
        letters = list(pattern)
        words = s.split(" ")

        if len(letters) != len(words):
            return False

        idx = 0
        letterDict = {}
        wordDict = {}

        while idx < len(letters):
            word = words[idx]
            letter = letters[idx]

            if (letter in letterDict) != (word in wordDict):
                return False

            if letter in letterDict and letterDict[letter] != word:
                return False

            if word in wordDict and wordDict[word] != letter:
                return False

            letterDict[letter] = word
            wordDict[word] = letter

            idx += 1

        return True
```

```python
class Solution:
    def reverseString(self, s: List[str]) -> None:
        n = len(s)

        idx_begin = 0
        idx_end = n - 1

        while idx_begin <= n // 2 - 1:
            temp = s[idx_begin]
            s[idx_begin] = s[idx_end]
            s[idx_end] = temp

            idx_begin += 1
            idx_end -= 1
```
