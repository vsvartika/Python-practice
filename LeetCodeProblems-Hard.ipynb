```python
# Merging of sorted arrays
#idea: take the median of the respective arrays, and discard half of it. Keep repeating
```

```python

def medianOfTwoSortedArrays(nums1,nums2):
    array1 = nums1
    array2 = nums2

    while len(array1)>1 or len(array2)>1:
        #idxs to be removed
        if array1[0]<= array2[0]:  # check the smallest entry
            array1 = array1[1:]
        else:
            array2 = array2[1:]   

        if array1[-1]>= array2[-1]:  # check the largest entry
            array1 = array1[:-1]
        else:
            array2 = array2[:-1]  

        if len(array1)>0 and len(array2)>0:
            return (array1[0] + array2[0])/2
        elif len(array1)>0:
            return array1[0]
        else:
            return array2[0]

nums1 = [1]
nums2 = [2,3]
#comnbined = [1,2,3,4,5,5,6,7]
medianOfTwoSortedArrays(nums1,nums2)
    
```
