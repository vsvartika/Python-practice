```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        if x<0:
            return False
        ls = []
        while True:
            element = x%10 #remainder
            remaining = x//10
            ls.append(element)
            if remaining ==0:
                break
            x = remaining

        if ls == ls[::-1]:
            return True
        else:
            return False

sol = Solution()
sol.isPalindrome(12021)
            
```

```python
def isValid(s: str) -> bool:
    paranthesis_dict = {'{':'}', '(':')', '[':']'}
    open_pattern = []

    if len(s)%2 == 1:
        return False
    
    Matched = True
    for char in s:          
        if char in paranthesis_dict:
            open_pattern.append(char)
        else:
            if len(open_pattern)>=1:
                prev_char = open_pattern.pop(-1)
                if paranthesis_dict[prev_char] != char:
                    return False     
            else:
                return False 
            
    if len(open_pattern)>0:
        return False                                                  
    return Matched

strs = "(("
isValid(strs)
```

```python
def romanToInt(s: str) -> int:
    roman_to_int = {'I':1, 'V':5, 'X':10, 'L':50, 'C':100, 'D':500, 'M':1000}
    exception_instances = {'I':('V','X'), 'X':('L','C'), 'C':('D','M')}
    exception_values = {'IV':4, 'IX':9, 'XL':40, 'XC':90, 'CD':400, 'CM':900}
    int_vals = []
    skip_this = False
    for idx,numeral in enumerate(s[:-1]):
        if not skip_this:
            if numeral in exception_instances and s[idx+1] in exception_instances[numeral]:
                skip_next = True
                int_vals.append(exception_values[numeral+s[idx+1]])
            else:
                skip_next = False
                int_vals.append(roman_to_int[numeral])
        else:
            skip_next = False
        skip_this = skip_next

    if not skip_this:
        int_vals.append(roman_to_int[s[-1]])  
    return sum(int_vals)      
s= "MCM"
romanToInt(s)   
```

```python
def strStr(haystack: str, needle: str) -> int:
    window = len(needle)
    idx_start = 0
    while idx_start + window <=len(haystack):
        if haystack[idx_start:idx_start+window] == needle:
            return idx_start
        else:
            idx_start = idx_start+1   
    return -1

haystack = "butsad"
needle = "sad"

strStr(haystack, needle)
```

```python
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def deleteDuplicates(head: ListNode) -> ListNode:
    current = head
    while current and current.next:
        if current.val == current.next.val:
            current.next = current.next.next
        else:
            current = current.next
    return head

node1 = ListNode(1)
node2 = ListNode(30)
node3 = ListNode(30)
node4 = ListNode(30)

node1.next = node2
node2.next = node3
node3.next = node4

cleanedList = deleteDuplicates(node1)
current = cleanedList
while current:
    print(current.val, end = "-->")
    current = current.next
```
