```python
class Solution:
    def isPalindrome(self, x: int) -> bool:
        if x<0:
            return False
        ls = []
        while True:
            element = x%10 #remainder
            remaining = x//10
            ls.append(element)
            if remaining ==0:
                break
            x = remaining

        if ls == ls[::-1]:
            return True
        else:
            return False

sol = Solution()
sol.isPalindrome(12021)
            
```

```python
def isValid(s: str) -> bool:
    paranthesis_dict = {'{':'}', '(':')', '[':']'}
    open_pattern = []

    if len(s)%2 == 1:
        return False
    
    Matched = True
    for char in s:          
        if char in paranthesis_dict:
            open_pattern.append(char)
        else:
            if len(open_pattern)>=1:
                prev_char = open_pattern.pop(-1)
                if paranthesis_dict[prev_char] != char:
                    return False     
            else:
                return False 
            
    if len(open_pattern)>0:
        return False                                                  
    return Matched

strs = "(("
isValid(strs)
```

```python
def romanToInt(s: str) -> int:
    roman_to_int = {'I':1, 'V':5, 'X':10, 'L':50, 'C':100, 'D':500, 'M':1000}
    exception_instances = {'I':('V','X'), 'X':('L','C'), 'C':('D','M')}
    exception_values = {'IV':4, 'IX':9, 'XL':40, 'XC':90, 'CD':400, 'CM':900}
    int_vals = []
    skip_this = False
    for idx,numeral in enumerate(s[:-1]):
        if not skip_this:
            if numeral in exception_instances and s[idx+1] in exception_instances[numeral]:
                skip_next = True
                int_vals.append(exception_values[numeral+s[idx+1]])
            else:
                skip_next = False
                int_vals.append(roman_to_int[numeral])
        else:
            skip_next = False
        skip_this = skip_next

    if not skip_this:
        int_vals.append(roman_to_int[s[-1]])  
    return sum(int_vals)      
s= "MCM"
romanToInt(s)   
```

```python
def maxProfit( prices: list[int]) -> int:
    max_sell_price = 0
    max_profit = 0
    best_day = 0
    for day in range(len(prices)-1,-1,-1):
        buy_price = prices[day]
        profit = max_sell_price - buy_price
        if profit >= max_profit:
            best_day = day
            max_profit = profit
        max_sell_price = max(max_sell_price,buy_price)
    return max_profit

        
data = [7,6,4,3,1]
maxProfit(data)
```

```python
#majority element

def majorityElement(nums: list[int])-> int:
    nums_dict = {}
    for num in nums:
        if num in  nums_dict.keys():
            nums_dict[num] += 1 
        else:
            nums_dict[num]=1

    freq = 0

    for key, value in nums_dict.items():
        if value> freq:
            maj_number = key
            freq = value

    return maj_number

nums = [3,2,2,3,3]
majorityElement(nums)
```

```python
#power of two
def isPowerOfTwo( n: int) -> bool:
    num = abs(n)
    prev_val = 1
    new_val = 2
    while True:
        if num <=0:
            return False
        if num == prev_val or num==new_val:
            return True
        elif num>prev_val and num<new_val:
            return False
        else:
            prev_val = new_val
            new_val = new_val*2

isPowerOfTwo(44)
```
